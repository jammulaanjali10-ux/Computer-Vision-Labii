```python
import cv2
import matplotlib.pyplot as plt

# Read image in grayscale
img = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)

# Define gray-level range
A = 100
B = 180

# Gray level slicing WITH background
with_background = img.copy()

with_background[(img >= A) & (img <= B)] = 255

# Gray level slicing WITHOUT background
without_background = img.copy()

without_background[(img >= A) & (img <= B)] = 255
without_background[(img < A) | (img > B)] = 0

# Display results
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(with_background, cmap='gray')
plt.title("With Background")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(without_background, cmap='gray')
plt.title("Without Background")
plt.axis("off")

plt.tight_layout()
plt.show()
```
