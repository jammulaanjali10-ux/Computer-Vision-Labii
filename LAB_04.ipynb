```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def show_comparison(original, sobel, prewitt, canny):
    plt.figure(figsize=(12, 10))

    # Original Image
    plt.subplot(2, 2, 1)
    plt.imshow(original, cmap='gray')
    plt.title("Original Grayscale Image")
    plt.axis('off')

    # Sobel
    plt.subplot(2, 2, 2)
    plt.imshow(sobel, cmap='gray')
    plt.title("Sobel Edge Detection")
    plt.axis('off')

    # Prewitt
    plt.subplot(2, 2, 3)
    plt.imshow(prewitt, cmap='gray')
    plt.title("Prewitt Edge Detection")
    plt.axis('off')

    # Canny
    plt.subplot(2, 2, 4)
    plt.imshow(canny, cmap='gray')
    plt.title("Canny Edge Detection")
    plt.axis('off')

    plt.tight_layout()
    plt.show()


# Read image
img = cv2.imread('img1.jpg', 0)

if img is None:
    print("Error loading image")

else:

    # ---------- SOBEL ----------
    sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
    sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

    sobel = cv2.magnitude(sobelx, sobely)
    sobel = np.uint8(np.absolute(sobel))


    # ---------- PREWITT ----------
    kernelx = np.array([[1, 1, 1],[0, 0, 0],[-1, -1, -1]], dtype=np.float32)
    kernely = np.array([[-1, 0, 1],[-1, 0, 1],[-1, 0, 1]], dtype=np.float32)

    prewittx = cv2.filter2D(img, cv2.CV_32F, kernelx)
    prewitty = cv2.filter2D(img, cv2.CV_32F, kernely)

    prewitt = cv2.magnitude(prewittx, prewitty)
    prewitt = np.uint8(np.absolute(prewitt))


    # ---------- CANNY ----------
    canny_edges = cv2.Canny(img, 100, 200)
    show_comparison(img,sobel, prewitt, canny_edges
  )
```
